```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Embedding, Conv1D, GlobalMaxPooling1D, Dense

# vocabulary size and maximum review length
max_words = 10000
seq_len = 200

# load imdb sentiment dataset
(train_x, train_y), (test_x, test_y) = imdb.load_data(num_words=max_words)

# pad sequences so that every review has same length
train_x = sequence.pad_sequences(train_x, maxlen=seq_len)
test_x = sequence.pad_sequences(test_x, maxlen=seq_len)

# build CNN based sentiment classifier
model = Sequential([
    Embedding(input_dim=max_words, output_dim=128, input_length=seq_len),
    Conv1D(128, 5, activation="relu"),
    GlobalMaxPooling1D(),
    Dense(10, activation="relu"),
    Dense(1, activation="sigmoid")
])

# compile the network
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# train the model
model.fit(train_x, train_y, epochs=5, batch_size=64)

# evaluate on test data
test_loss, test_acc = model.evaluate(test_x, test_y)

print("Test Accuracy:", test_acc)
```
