```python
import numpy as np
import pandas as pd
import yfinance as yf
from sklearn.preprocessing import MinMaxScaler
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

# stock symbol
symbol = "AAPL"

# download historical price data
df = yf.download(symbol, start="2015-01-01", end="2024-01-01")

# use closing price only
close_vals = df["Close"].values.reshape(-1, 1)

# scale prices between 0 and 1
scaler = MinMaxScaler(feature_range=(0, 1))
scaled_vals = scaler.fit_transform(close_vals)

window = 60

X_data = []
y_data = []

# create time-series sequences
for i in range(window, len(scaled_vals)):
    X_data.append(scaled_vals[i-window:i])
    y_data.append(scaled_vals[i])

X_data = np.array(X_data)
y_data = np.array(y_data)

# split train and test sets
cut = int(len(X_data) * 0.8)

X_train = X_data[:cut]
y_train = y_data[:cut]

X_test = X_data[cut:]
y_test = y_data[cut:]

# build LSTM model
model = Sequential()

model.add(LSTM(50, return_sequences=True, input_shape=(window, 1)))
model.add(LSTM(50))
model.add(Dense(1))

# compile model
model.compile(optimizer="adam", loss="mean_squared_error")

# train model
model.fit(X_train, y_train, epochs=10, batch_size=32)

# prediction on test data
predictions = model.predict(X_test)

# convert back to original price scale
predictions = scaler.inverse_transform(predictions)

print(predictions[:10])
```
