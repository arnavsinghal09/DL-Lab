```python
import numpy as np
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# vocabulary size and max review length
max_words = 10000
review_len = 200

# load IMDB dataset
(train_x, train_y), (test_x, test_y) = imdb.load_data(num_words=max_words)

# make all sequences equal length
train_x = sequence.pad_sequences(train_x, maxlen=review_len)
test_x = sequence.pad_sequences(test_x, maxlen=review_len)

# build the RNN model
model = Sequential()
model.add(Embedding(input_dim=max_words, output_dim=128, input_length=review_len))
model.add(SimpleRNN(64))
model.add(Dense(1, activation="sigmoid"))

# compile model
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# train the network
model.fit(train_x, train_y, epochs=5, batch_size=64)

# evaluate performance
test_loss, test_acc = model.evaluate(test_x, test_y)

print("Evaluation Results")
print("Test Accuracy:", test_acc)
```
